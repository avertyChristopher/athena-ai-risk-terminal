# 01 - Market Data Returns

Exploration notebook for price returns and volatility calculations used by the backend domain layer.

```python
from math import sqrt

prices = [100.0, 101.5, 99.8, 103.2, 104.1, 102.7, 106.4]
simple_returns = [current / previous - 1 for previous, current in zip(prices, prices[1:])]
simple_returns
```

```python
def sample_stdev(values):
    mean = sum(values) / len(values)
    variance = sum((value - mean) ** 2 for value in values) / (len(values) - 1)
    return sqrt(variance)

daily_vol = sample_stdev(simple_returns)
annualized_vol = daily_vol * sqrt(252)

{'daily_volatility': daily_vol, 'annualized_volatility': annualized_vol}
```

```python
window = 3
rolling_volatility = [
    sample_stdev(simple_returns[index - window:index])
    for index in range(window, len(simple_returns) + 1)
]
rolling_volatility
```
